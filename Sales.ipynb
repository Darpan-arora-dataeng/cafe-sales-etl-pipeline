# Milestone 4 - Data Cleaning

## Initialize the working DataFrame

```python
df = raw_df.copy()
```

## 4.1 Convert Numeric Columns

```python
numeric_cols = ["Quantity", "Price Per Unit", "Total Spent"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.astype({
    "Quantity": "Int64",
    "Price Per Unit": "float64",
    "Total Spent": "float64"
})
```

```python
df[numeric_cols].dtypes
```

```python
df[numeric_cols].isna().sum()
```

## 4.2 Recover missing quantity Values

```python
# Use the business rule:

# Quantity = Total Spent / Price Per Unit

recoverable_qty = (
    df["Quantity"].isna()
    & df["Price Per Unit"].notna()
    & df["Total Spent"].notna()
    & df["Price Per Unit"].gt(0)
)

recoverable_qty.sum()
```

```python
calculated_qty = (
    df.loc[recoverable_qty, "Total Spent"]
    / df.loc[recoverable_qty, "Price Per Unit"]
)

non_whole_qty = ~np.isclose(
    calculated_qty,
    calculated_qty.round()
)

non_whole_qty.sum()
```

```python
if non_whole_qty.any():
    raise ValueError("Some calculated quantities are not whole numbers.")

df.loc[recoverable_qty, "Quantity"] = (
    calculated_qty.round().astype("Int64")
)

df["Quantity"].isna().sum()
```

## 4.3 Recover Missing Price Per Unit Values

```python
# Use the business rule:

# Price Per Unit = Total Spent / Quantity

recoverable_price = (
    df["Price Per Unit"].isna()
    & df["Quantity"].notna()
    & df["Total Spent"].notna()
    & df["Quantity"].gt(0)
)

recoverable_price.sum()
```

```python
calculated_price = (
    df.loc[recoverable_price, "Total Spent"]
    / df.loc[recoverable_price, "Quantity"]
)

calculated_price.describe()
```

```python
df.loc[recoverable_price, "Price Per Unit"] = calculated_price

df["Price Per Unit"].isna().sum()
```

## 4.4 Recover Missing Total Spent Values

```python
# Use the business rule:

# Total Spent = Quantity × Price Per Unit

recoverable_total = (
    df["Total Spent"].isna()
    & df["Quantity"].notna()
    & df["Price Per Unit"].notna()
    & df["Quantity"].gt(0)
)

recoverable_total.sum()
```

```python
calculated_total = (
    df.loc[recoverable_total, "Quantity"]
    * df.loc[recoverable_total, "Price Per Unit"]
)

calculated_total.describe()
```

```python
df.loc[recoverable_total, "Total Spent"] = calculated_total

df["Total Spent"].isna().sum()
```

## 4.5 Standardize/Clean the Item column

```python
df["Item"] = df["Item"].replace({
    "ERROR": "Unknown",
    "UNKNOWN": "Unknown"
})

df["Item"] = df["Item"].fillna("Unknown")

df["Item"].value_counts(dropna=False)
```

## 4.6 Standardize/Clean categorical columns(Payment Method and Location)

```python
text_cols = ["Payment Method", "Location"]

for col in text_cols:
    df[col] = df[col].replace(["ERROR", "UNKNOWN"], "Unknown")
    df[col] = df[col].fillna("Unknown")

for col in text_cols:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

## 4.7 Convert transaction dates

```python
df["Transaction Date"] = df["Transaction Date"].replace(
    ["ERROR", "UNKNOWN"], np.nan
)

df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    format="%m/%d/%Y",
    errors="coerce"
)

print("Data type:", df["Transaction Date"].dtype)
print("Missing dates:", df["Transaction Date"].isna().sum())
```

## 4.8 Set Text Column Data Types

```python
df["Transaction ID"] = df["Transaction ID"].astype("string")

text_cols = ["Item", "Payment Method", "Location"]
df[text_cols] = df[text_cols].astype("string")

df.dtypes
```

# Milestone 5 - Data Transformation

## 5.1 Create a Transformation Dataset

```python
transformed_df = df.copy()

transformed_df.shape
```

## 5.2 Create Revenue Column

```python
transformed_df["Revenue"] = transformed_df["Total Spent"]

transformed_df[["Total Spent", "Revenue"]].head(10)
```

## 5.3 Extract Month from Transaction Date

```python
transformed_df["Month"] = transformed_df["Transaction Date"].dt.month

transformed_df[["Transaction Date", "Month"]].head(10)
```

## 5.4 Extract Weekday from Transaction Date

```python
transformed_df["Weekday"] = (
    transformed_df["Transaction Date"].dt.day_name()
)

transformed_df[["Transaction Date", "Weekday"]].head(10)
```

## 5.5 Extract Year from Transaction Date

```python
transformed_df["Year"] = transformed_df["Transaction Date"].dt.year

transformed_df[["Transaction Date", "Year"]].head(10)
```

## 5.6 Create Year-Month Column

```python
transformed_df["Year_Month"] = (
    transformed_df["Transaction Date"].dt.to_period("M").astype("string")
)

transformed_df[["Transaction Date", "Year_Month"]].head(10)
```

## 5.7 Create Item Category Mapping

```python
category_mapping = {
    "Coffee": "Beverage",
    "Tea": "Beverage",
    "Juice": "Beverage",
    "Smoothie": "Dessert",
    "Salad": "Starter",
    "Cake": "Dessert",
    "Sandwich": "Starter",
    "Cookie": "Dessert",
    "Unknown": "Unknown"
}

transformed_df["Item_Category"] = transformed_df["Item"].map(category_mapping)

transformed_df[["Item", "Item_Category"]].value_counts(dropna=False)
```

## 5.8 Validate Transformed Dataset

```python
transformed_df.shape
```

```python
transformed_df.columns
```

```python
# Check unmapped categories

transformed_df.loc[
    transformed_df["Item_Category"].isna(),
    "Item"
].value_counts(dropna=False)
```

```python
# Convert Month and Year to nullable integers
transformed_df["Month"] = transformed_df["Month"].astype("Int64")
transformed_df["Year"] = transformed_df["Year"].astype("Int64")

# Convert text columns to Pandas string dtype
text_cols = ["Weekday", "Item_Category"]

for col in text_cols:
    transformed_df[col] = transformed_df[col].astype("string")

# Verify the updated data types
transformed_df.dtypes
```

## 6.1 Prepare DataFrame for PostgreSQL

```python
# Create a copy so transformed_df remains unchanged

load_df = transformed_df.copy()

# Rename columns to match the PostgreSQL table
load_df = load_df.rename(
    columns={
    "Transaction ID": "transaction_id",
    "Item": "item",
    "Quantity": "quantity",
    "Price Per Unit": "price_per_unit",
    "Total Spent": "total_spent",
    "Payment Method": "payment_method",
    "Location": "location",
    "Transaction Date": "transaction_date",
    "Revenue": "revenue",
    "Month": "month",
    "Weekday": "weekday",
    "Year": "year",
    "Year_Month": "year_month",
    "Item_Category": "item_category"
})

load_df.head()
```

## 6.3 Connect Python to PostgreSQL

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine, URL

load_dotenv()

connection_url = URL.create(
    drivername="postgresql+psycopg",
    username=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    host=os.getenv("DB_HOST"),
    port=int(os.getenv("DB_PORT", "5432")),
    database=os.getenv("DB_NAME")
)

engine = create_engine(connection_url)
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(text("SELECT current_database();"))
    print(result.scalar())
```

## 6.4 Load data into PostgreSQL

```python
load_df.to_sql(
    name="cafe_sales",
    con=engine,
    schema="public",
    if_exists="append",
    index=False,
    chunksize=1000
)
```

## 6.5 Verify the data in PostgreSQL

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(
        text("SELECT COUNT(*) FROM public.cafe_sales;")
    )
    print("Rows in PostgreSQL:", result.scalar())
```

## 6.6 Validate the loaded data

```python
with engine.connect() as connection:
    db_count = connection.execute(
        text("SELECT COUNT(*) FROM public.cafe_sales;")
    ).scalar()

print("Pandas rows:", len(load_df))
print("PostgreSQL rows:", db_count)
print("Row counts match:", len(load_df) == db_count)
```

```python
null_check_query = text("""
    SELECT
        COUNT(*) FILTER (WHERE quantity IS NULL) AS missing_quantity,
        COUNT(*) FILTER (WHERE price_per_unit IS NULL) AS missing_price,
        COUNT(*) FILTER (WHERE total_spent IS NULL) AS missing_total,
        COUNT(*) FILTER (WHERE transaction_date IS NULL) AS missing_date,
        COUNT(*) FILTER (WHERE revenue IS NULL) AS missing_revenue
    FROM public.cafe_sales;
""")

with engine.connect() as connection:
    result = connection.execute(null_check_query).mappings().one()
    print(dict(result))
```
